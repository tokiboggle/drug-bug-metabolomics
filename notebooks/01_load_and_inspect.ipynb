## Load the files


```python
from pathlib import Path

import pandas as pd

# Works whether you launch the notebook from the repo root or from notebooks/
cwd = Path.cwd().resolve()
candidates = [cwd / "data" / "raw", cwd.parent / "data" / "raw"]
data_dir = next((p for p in candidates if p.exists()), None)
if data_dir is None:
    raise FileNotFoundError(f"Could not find data/raw from {cwd}")

print(f"Reading files from: {data_dir}")

metadata = pd.read_csv(data_dir / "group2_drugs_metadata.csv")
features = pd.read_csv(data_dir / "group2_drugs_iimn_gnps_quant.csv")
annotations = pd.read_csv(data_dir / "group2_drugs_annotations.csv")
```

## A first look


```python
print("metadata:", metadata.shape[0], "rows ×", metadata.shape[1], "columns")
print("columns:", list(metadata.columns))
metadata.head()
```

```python
print("features:", features.shape[0], "rows ×", features.shape[1], "columns")
print("columns (first 15):", list(features.columns[:15]))
features.head()
```

```python
print("annotations:", annotations.shape[0], "rows ×", annotations.shape[1], "columns")
print("columns:", list(annotations.columns))
annotations.head()
```

Sample counts by type, from the metadata file.


```python
print("mzmine_sample_type")
print(metadata["mzmine_sample_type"].value_counts(dropna=False))
print()
print("ATRRIBUTE_SampleType")
print(metadata["ATRRIBUTE_SampleType"].value_counts(dropna=False))
print()
print("ATTRIBUTE_Drug")
print(metadata["ATTRIBUTE_Drug"].value_counts(dropna=False))
print()
print("ATTRIBUTE_Bacteria")
print(metadata["ATTRIBUTE_Bacteria"].value_counts(dropna=False))
print()
print("Bacterial samples (ATTRIBUTE_Bacteria == 'Yes'): treatment × time point")
bacterial = metadata[metadata["ATTRIBUTE_Bacteria"] == "Yes"]
print(
    pd.crosstab(
        bacterial["ATRRIBUTE_SampleType"],
        bacterial["ATTRIBUTE_Drug"],
        margins=True,
    )
)
```
